## BeautifulSoup Demo

BeautifulSoup works on static HTML only. For dynamic sites like Flipkart, Instagram, Twitter — we need Selenium or APIs.

Create a demo to fetch and parse HTML.

```python
import requests
from bs4 import BeautifulSoup

# Fetch web page
url = 'https://httpbin.org/html'
response = requests.get(url)

# Parse HTML
soup = BeautifulSoup(response.text, 'html.parser')

# Extract h1 tag
h1_tag = soup.find('h1')
print(h1_tag.text)
```

## API Demo

An API gives structured data directly in JSON format, so we don't need to scrape HTML. This is how Twitter, YouTube, Instagram, Reddit actually work.

```python
import requests

# API call
response = requests.get('https://jsonplaceholder.typicode.com/posts/1')
data = response.json()
print(data)
```
